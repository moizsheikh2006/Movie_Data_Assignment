### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns

drama_action = data[data.Genre.str.contains("Drama|Action", na=False)].copy()

drama_action["Genre_Type"] = drama_action.Genre.apply(
    lambda x: "Drama" if "Drama" in x else "Action"
)

sns.boxplot(data=drama_action, x="Genre_Type", y="Gross")
plt.title("Gross Revenue: Drama vs Action")
plt.xlabel("Genre")
plt.ylabel("Gross Revenue")
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import matplotlib.pyplot as plt

top5 = data.nlargest(5, "IMDB_Rating").copy()

top5["Rating_100_Scale"] = top5.IMDB_Rating * 10

top5.plot(
    x="Series_Title",
    y=["Meta_score", "Rating_100_Scale"],
    kind="bar",
    figsize=(12, 6),
    width=0.8
)

plt.title("Top 5 Movies: Meta Score vs IMDb Rating")
plt.xlabel("Movie Name")
plt.ylabel("Score (Out of 100)")
plt.xticks(rotation=45, ha="right")
plt.legend(["Meta Score", "IMDb Rating (Scaled to 100)"])
plt.tight_layout()
plt.show()
```
